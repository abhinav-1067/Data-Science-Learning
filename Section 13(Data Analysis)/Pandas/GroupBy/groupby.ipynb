```python
import pandas as pd
import numpy as np
```

```python
movie = pd.read_csv('imdb-top-1000.csv')
movie.head(3)
```

```python
## finding director with  most popularity,

movie.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# find the highest rated movie in each genre,

movie.groupby(['Series_Title','Genre'])['IMDB_Rating'].max().sort_values(ascending=False)
```

```python
## find number of movies done by each actors,

movie.groupby('Star1')['Series_Title'].count()
```

```python
## we can also find that after grouping how many numbers of group is created,

len(movie.groupby('Genre'))
```

```python
movie['Genre'].nunique()
```

```python
# finding the number of movies that starts with letter A for each group,

genres = movie.groupby('Genre')

def fin(group):
    return print(group['Series_Title'].str.startswith('A').sum())

genres.apply(fin)
```
