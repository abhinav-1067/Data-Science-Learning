```python
##

cc = pd.merge(courses,nov,on='course_id',how='left')
cc['course_name'].value_counts()
```

```python
reg = pd.concat([nov,dec], ignore_index=True)
```

```python
## find the total revenue generated,

total_revenue = (pd.merge(reg,courses,on='course_id',how='inner'))
int(total_revenue['price'].sum())
```

```python
## find the revenue generated month by month,

for_nov = pd.merge(nov,courses,on='course_id',how='inner')['price'].sum()
for_dec = pd.merge(dec,courses, on='course_id',how='inner')['price'].sum()

print("Total revenue in November: ",for_nov)
print("Total revenue in December: ",for_dec)
```

```python
## another way for this question,,

temp_df = pd.concat([nov,dec],keys=['NOV','DEC']).reset_index()
rev = pd.merge(temp_df,courses,on='course_id').groupby('level_0')['price'].sum()
rev
```

```python
## Print the registration table,,
# columns --> name -> course -> price.

reg_table = pd.merge(reg,students,on='student_id')
dd = reg_table.merge(courses,on='course_id')
del dd['partner']
dd
```
