```python
import numpy as np
import pandas as pd
```

### Series from lists

```python
# string

country = ['India','USA','Srilanka']
pd.Series(country)
```

```python
# intergers

score = [12,3,2,3,2,53]
pd.Series(score)
```

```python
# custom indexes

marks = [23,66,39,77]
subjects = ['Maths','English','Physics','Chemistry']

pd.Series(marks, index=subjects)
```

```python
# series with dictionary

s = {
    'maths':27,
    'english':88,
    'Science': 56,
    'Hindi':98
}

marks = pd.Series(s)
marks
```

### Pandas Series Attributes

```python
s = pd.Series([10,20,30,40,None], index=['a','b','c','d','e'])

# Attributes

print("Index of the Series: ",s.index)
print("Values In the Series: ",s.values)
print("Data type of the Series: ",s.dtype)
print("Size of the series: ",s.size)
print("Shape of the Series: ",s.shape)
print("check NaN values in Series: ",s.hasnans)
print("Check that Series has unqiue values or not: ",s.is_unique)
print("check number of Dimensions of Series: ",s.ndim)
print("Check that Series is empty or not: ",s.empty)
```
