```python
##code evaluating speed,

# in python list

import time

a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]

c = []

start = time.time()
for i in range(len(a)):
    c.append(a[i] + b[i])
print(time.time() - start)
```

```python
# now code evaluating speed in numpy
import numpy as np

aa = np.arange(10000000)
bb = np.arange(10000000,20000000)

start = time.time()
c = aa + bb
print(time.time() - start)
```

```python
# now we are comparing that how much memory does python list and numpy list uses,

import sys
ac = [i for i in range(10000000)]
sys.getsizeof(ac)
```

```python
# for numpy
av = np.arange(10000000)
sys.getsizeof(av)
```

```python
# can also change the dtyoe to reduce memory usage

av = np.arange(10000000, dtype=np.int32)
sys.getsizeof(av)
```
