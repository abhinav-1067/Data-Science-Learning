```python
import pandas as pd
```

```python
df = pd.read_csv('startup_funding.csv')
```

```python
df.head(3)
```

```python
## Dropping Remarks Column,

df.drop(columns=['Remarks'],inplace=True)
```

```python
## Setting Sr No as Index,

df.set_index('Sr No')
```

```python
df.info()
```

```python
# Rename columns,
df.rename(columns={
    'Date dd/mm/yyyy':'Date',
    'Startup Name':'Startups',
    'City  Location':'City',
    'Investors Name':'Investors',
    'InvestmentnType':'Investment Type',
    'Amount in USD':'Amount'
}, inplace=True)
```

```python
df['Amount'].dtype
```

```python
df['Amount'] = df['Amount'].str.replace(',','',regex=False)
```

```python
df
```

```python
df.info()
```

```python
## first dropping all NaN rows,

df = df.dropna(subset=['Date','Startups','Industry Vertical','SubVertical','City','Investors','Investment Type','Amount'])
```

```python
df
```

```python
df = df[df['Amount'].str.isdigit()]
```

```python
df['Amount'] = df['Amount'].astype('float')
```

```python
df
```

```python
df.info()
```

```python
def to_inr(dollar):
    inr = dollar * 95.8
    return inr/10000000
```

```python
df['Amount'] = df['Amount'].apply(to_inr)
```

```python
df
```

```python
# converting Date colum to DateTime object,

df['Date'] = pd.to_datetime(df['Date'],errors='coerce')
```

```python
df.info()
```

```python
df.to_csv('startup_cleaned.csv', index=False)
```
